## 7) Análisis de complejidad **[5 pts]**

### 1. Justificación
El análisis de complejidad temporal y espacial se realiza evaluando cada una de las operaciones clave del sistema de gestión de la tienda, las cuales se dividen según la estructura de datos utilizada:

* **Cola de Clientes (FIFO):**
  * **Agregar al final (`add_users_h` / opción 1):** Utiliza el método `.append()`, el cual tiene un tiempo constante en promedio $O(1)$.
  * **Atender al cliente (`attend` / opción 2):** Utiliza `.pop(0)` sobre una lista de Python para extraer el primer elemento. Esto desplaza todos los elementos restantes una posición a la izquierda, tomando un tiempo lineal $O(n)$, donde $n$ es el número de elementos en la cola.
  * **Deshacer atención (`move_forward` / opción 5):** Utiliza `.insert(0, name)` para devolver al cliente al inicio de la cola. Al igual que el `pop(0)`, requiere desplazar los elementos, resultando en una complejidad temporal de $O(n)$.

* **Pila de Historial de Atenciones (LIFO):**
  * **Apilar y desapilar (`pile` / `unpile`):** Las operaciones se realizan al final de la estructura mediante `.append()` y `.pop()`, lo cual toma un tiempo constante $O(1)$.

* **Catálogo de Productos (Lista Doblemente Enlazada simulada con diccionarios y listas):**
  * **Agregar producto (`add_product`):** Inserta al final de la lista utilizando `.append()`, con un costo de $O(1)$.
  * **Buscar y Eliminar producto (`search_product` / `delete_product`):** Requiere recorrer la lista linealmente para coincidir con el código del producto, lo que genera una complejidad en el peor caso de $O(n)$, donde $n$ es el número total de productos registrados.


### 2. Función $T[n]$ General del Sistema
Para la operación de la cola (que involucra el desplazamiento de elementos), la función de tiempo de ejecución en función del tamaño de la entrada $n$ (número de elementos) se modela mediante un polinomio de primer grado:

$$T[n] = c_1 \cdot n + c_2$$

Donde:
* $c_1$ representa el costo por cada desplazamiento unitario en la lista al insertar o extraer del inicio.
* $c_2$ representa las operaciones de tiempo constante (asignaciones, validaciones de menú e impresión).

Por lo tanto, la complejidad asintótica dominante es **$O(n)$** para las operaciones de cola basadas en índices iniciales, y **$O(1)$** para las operaciones de pila y adición al final.


### 3. Mejor y Peor Caso

* **Mejor Caso $\Omega(1)$:** 
  * Ocurre en operaciones de pila (apilar/desapilar el último atendido), inserciones al final de la cola, o cuando se busca/elimina un producto que se encuentra exactamente en la primera posición de la lista del catálogo. El tiempo de ejecución es independiente del tamaño de la entrada.
* **Peor Caso $O(n)$:** 
  * Ocurre al **atender un cliente** o **deshacer una atención**, ya que Python debe reorganizar y desplazar toda la cola de tamaño $n$. Asimismo, se presenta al buscar o eliminar un producto que está ubicado al final del catálogo o que no existe, obligando a recorrer el 100% de los elementos.



## 8) Código funcional **[10 pts]**

```python
''' USERS LINE FUNCTIONS '''
def users_line_empty(users_line):
    return len(users_line) == 0

def add_users_h(users_line, name):
    users_line.append(name)

def move_forward(users_line, name):
    users_line.insert(0, name)

def attend(users_line):
    if users_line_empty(users_line):
        return None
    return users_line.pop(0)

def show_users_line(users_line):
    if users_line_empty(users_line):
        print("\n\t\tNo hay clientes esperando.")
        return
    print("\n\t\tClientes esperando: " + " -> ".join(users_line))

''' USERS HISTORY FUCTIONS'''
def users_history_empty(users_history):
    return len(users_history) == 0

def pile(users_history, name):
    users_history.append(name)

def unpile(users_history):
    if users_history_empty(users_history):
        return None
    return users_history.pop()

def show_last(users_history):
    if users_history_empty(users_history):
        return None
    return users_history[-1]

def show_history(users_history):
    if users_history_empty(users_history):
        print("\n\t\tAún no se ha atendido a ningún cliente.")
        return
    print("\n\tHistorial de atención (tope -> fondo): " + " -> ".join(reversed(users_history)))

'''PRODUCTS FUCTIONS'''
def products_is_empty(products):
    return len(products) == 0

def add_product(products, code, name, price):
    products.append({"código": code, "nombre": name, "precio": price})

def go_forward(products):
    if products_is_empty(products):
        print("\n\t\tEl catálogo está vacío.")
        return
    names = [product["nombre"] for product in products]
    print("\t\tNULL <- " + " <-> ".join(names) + " -> NULL")

def go_back(products):
    if products_is_empty(products):
        print("\n\t\tEl catálogo está vacío.")
        return
    names = [product["nombre"] for product in reversed(products)]
    print("NULL <- " + " <-> ".join(names) + " -> NULL")

def search_product(products, code):
    for product in products:
        if product["código"] == code:
            return product
    return None

def delete_product(products, code):
    product = search_product(products, code)
    if product is None:
        return False
    products.remove(product)
    return True

'''SYSTEM'''

def ask_int_num(mensaje):
    while True:
        num = input(mensaje).strip()
        if num.isdigit():
            return int(num)
        print("\n\t\tPor favor ingrese un número válido.")

def ask_float_num(mensaje):
    while True:
        num = input(mensaje).strip()
        try:
            return float(num)
        except ValueError:
            print("\n\t\tPor favor ingrese un número válido.")

def start_users(users_line):
    n = ask_int_num("\n\t¿Cuántos clientes desea registrar?: ")
    for i in range(1, n + 1):
        name = input(f"\n\tIngrese el nombre del cliente {i}: ").strip()
        add_users_h(users_line, name)

def start_products(products):
    print("\n\tCarga inicial del catálogo de productos (puede dejar vacío para no cargar ninguno).")
    n = ask_int_num("\n\t¿Cuántos productos desea registrar?: ")
    for i in range(1, n + 1):
        print(f"\n\t\t--- Producto {i} ---")
        code = ask_int_num("\n\t\tCódigo: ")
        name = input("\t\tNombre: ").strip()
        price = ask_float_num("\t\tPrecio: ")
        add_product(products, code, name, price)

def users_menu(users_line, users_history):
    while True:
        print("\n\t\t--- MENÚ CLIENTES (Cola / Pila) ---")
        print("\t1. Agregar un nuevo cliente a la fila/cola de usuarios")
        print("\t2. Atender al siguiente cliente")
        print("\t3. Mostrar clientes esperando")
        print("\t4. Mostrar último cliente atendido")
        print("\t5. Deshacer última atención")
        print("\t6. Mostrar historial completo de atenciones")
        print("\t0. Volver al menú principal")
        option = input("\n\t\tSeleccione una opción: ").strip()

        if option == "1":
            name = input("\n\tIngrese el nuevo cliente: ").strip()
            add_users_h(users_line, name)
            show_users_line(users_line)

        elif option == "2":
            name = attend(users_line)
            if name is None:
                print("\n\t\tNo hay clientes para atender.")
            else:
                pile(users_history, name)
                print(f"\n\tAtendiendo a: {name}")
                show_users_line(users_line)

        elif option == "3":
            show_users_line(users_line)

        elif option == "4":
            last = show_last(users_history)
            if last is None:
                print("\n\t\tAún no se ha atendido a ningún cliente.")
            else:
                print(f"\n\t\tÚltimo cliente atendido: {last}")

        elif option == "5":
            name = unpile(users_history)
            if name is None:
                print("\n\t\tNo hay atenciones para deshacer.")
            else:
                move_forward(users_line, name)
                print(f"\n\t\tSe deshizo la atención de: {name}")
                print(f"\n\t{name} regresó al inicio de la fila/cola de usuarios.")
                show_users_line(users_line)

        elif option == "6":
            show_history(users_history)

        elif option == "0":
            break
        else:
            print("\n\t\tOpción inválida.")

def products_menu(products):
    while True:
        print("\n\t\t--- MENÚ CATÁLOGO DE PRODUCTOS ---\n")
        print("\t1. Agregar producto al final")
        print("\t2. Mostrar productos (primero -> último)")
        print("\t3. Mostrar productos (último -> primero)")
        print("\t4. Buscar producto por código")
        print("\t5. Eliminar producto por código")
        print("\t0. Volver al menú principal")
        option = input("\n\t\tSeleccione una opción: ").strip()

        if option == "1":
            code = ask_int_num("\n\t\tCódigo: ")
            name = input("\t\tNombre: ").strip()
            price = ask_float_num("\t\tPrecio: ")
            add_product(products, code, name, price)
            print("\n\t\t\t-- Producto agregado ---")
            go_forward(products)

        elif option == "2":
            go_forward(products)

        elif option == "3":
            go_back(products)

        elif option == "4":
            code = ask_int_num("\n\t\tIngrese el código a buscar: ")
            product = search_product(products, code)
            if product is None:
                print(f"\n\tNo se encontró ningún producto con el código {code}.")
            else:
                print(f"{product['código']} - {product['nombre']} - ${product['precio']:,.0f}")

        elif option == "5":
            code = ask_int_num("\n\t\tIngrese el código que desea eliminar: ")
            product = search_product(products, code)
            if product is None:
                print(f"\n\tNo se encontró ningún producto con el código {code}.")
            else:
                name_removed = product["nombre"]
                delete_product(products, code)
                print(f"\n\tDespués de eliminar el producto {name_removed}, la lista queda:")
                go_forward(products)

        elif option == "0":
            break
        else:
            print("\n\t\tOpción inválida.")

''' MAIN SYSTEM '''
def main():
    print("\n\t\t=== SISTEMA DE ATENCIÓN DE UNA TIENDA ===\n")

    users_line = []
    users_history = []
    products = []

    start_users(users_line)
    start_products(products)

    while True:
        print("\n\t\t===== MENÚ PRINCIPAL =====")
        print("\t1. Gestión de clientes")
        print("\t2. Gestión de catálogo de productos")
        print("\t0. Salir")
        option = input("\t\nSeleccione una opción: ").strip()

        if option == "1":
            users_menu(users_line, users_history)
        elif option == "2":
            products_menu(products)
        elif option == "0":
            print("\n\t\tCerrando programa...")
            break
        else:
            print("\n\t\tOpción inválida.")

if __name__ == "__main__":
    main()
```
