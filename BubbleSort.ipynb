## Enunciado
**Ejercicio: Organización de precios en una tienda**

Una tienda necesita organizar los precios de sus productos. La cantidad de productos puede variar, por lo que el usuario deberá indicar cuántos precios desea registrar.

El programa debe solicitar:

1. La cantidad de productos, representada por n.
2. El precio de cada uno de los productos.
3. El tipo de ordenamiento que se desea mostrar.
   
Después de ingresar los datos, el programa deberá utilizar el algoritmo Insertion Sort para organizar los precios.

**Punto 1. Orden ascendente:**
Ordenar los precios desde el menor hasta el mayor, con el fin de mostrar primero los productos más económicos.

**Punto 2. Orden descendente:**
Ordenar los precios desde el mayor hasta el menor, con el fin de mostrar primero los productos más costosos.

# Especificacion



**Historia del usuario(HU):**
El usuario ingresa el precio de un producto para que el valor quede guardado de manera correcta en el sistema para poder procesarlo

**Entrada :**
* Cantidad de productos ($n$)
* Precio de cada producto
* El tipo de ordenamiento que se desea hacer ($ascendente$ o $descendente$)

**Salida:**
Los precios organizados de forma ascendente o descente respectivamente a lo que el usuario haya escogido

**Requerimiento funcional:** La tarea que ejecutará el sistema


**Restricciones:**
Precio de productos: La entrada tiene que ser datos númericos, el valor del precio tiene que ser mayor que 0


# Diseño

El objetivo de este programa es facilitar la organizacion de precios de productos de forma rapida y confiable

Usamos un algoritmo de ordenamiento (Bubble Sort), ya que funciona comparando pares de precios cercanos y cambiandolos de posicion si no estan en el orden correcto.

La solucion es iterativa (se utilizan bucles for y while), creamos un proceso incremental ya que primero se recolectan todos los precios validos en una lista y despues se procesa la lista completamente para transformarla y ordenarla.

Para asegurar que el programa no falle ni entregue datos incorrectos, se mantienen dos condiciones durante la ejecución:

**Invariante de entrada:** Todos los datos dentro de la lista de precios siempre serán números estrictamente mayores a cero (p > 0).

**Invariante de ordenamiento:** Al finalizar cada ciclo externo del algoritmo, los elementos al final de la lista quedan en su posición definitiva y ordenada.

Esta solución permite organizar los datos hacia arriba o hacia abajo según la necesidad del usuario, al poder validarse las entradas evitamos que un número no valido dañe los cálculos, también usamos lógica intuitiva para no añadir complejidad innecesaria.

**Diagrama de flujo**



# Codigo

```python
def precio(prod):
    lista_p = []
    for i in range(prod):
        print("Producto #", i + 1)
        while True:
          p = float(input("Ingrese el precio del producto (mayor a cero): "))
          if p > 0:
            break
          print("Valor inválido, vuelva a ingresar el precio.")
        lista_p.append(p)
    return lista_p

def opcion(opc, lista_p):
    if opc == 1:
        for i in range(len(lista_p)):
            for j in range(len(lista_p)-1):
                if  lista_p[j] > lista_p[j+1]:
                    tempo = lista_p[j]
                    lista_p[j] = lista_p[j+1]
                    lista_p[j+1] = tempo
    else:
         for i in range(len(lista_p)):
            for j in range(len(lista_p)-1):
                if  lista_p[j] < lista_p[j+1]:
                    tempo = lista_p[j]
                    lista_p[j] = lista_p[j+1]
                    lista_p[j+1] = tempo
    return lista_p

def main():
    prod = int(input("Ingrese cuantos productos desea registrar: "))
    lista_p = precio(prod)
    while True:
        print("Si desea el ordenamiento ascendente ingrese 1 o descendente ingrese 2")
        opc = int(input("Opción: "))
        if opc == 1 or opc == 2:
            break
        print("Valor inválido, vuelva a ingresar una opción.")
    new_list = opcion(opc, lista_p)
    print ("Lista de precios ordenada según la opción escogida:", new_list)

if __name__ == "__main__":
    main()
```

#### Código base para benchmarking

```python
# Calcula el tiempo de ejecucion con base a diferentes tamaños de la entrada
# y nos da una lista de resultados, cada resultado contiene el tamaño de la entrada
# y el tiempo de ejecucion
# TENER EN CUENTA ESTAS ENTRADAS PUEDEN VARIAR SEGÚN EL PROBLEMA


def bench(n: int, step: int = 1) -> list[tuple[int, float]]:
    print("Generando datos ...", end='')
    from random import randint
    import time
    results: list[tuple[int, float]] = []


    # AJUSTAR LAS ENTRADAS SEGÚN EL PROBLEMA
    for size in range(1, n, step):
        input = size

        ti = time.perf_counter()
        output = input # Ajustar a su función
        tf = time.perf_counter()


        delta = (tf - ti) * 1000
        print(".", end="")
        results.append((size, delta))
    return results


# === CODIGO PARA DIBUJAR ===


def dibujar(x, y, dimensiones=(7, 5)):
    assert len(x) == len(y)
    import matplotlib.pyplot as plt
    plt.figure(figsize=dimensiones)
    plt.plot(x, y)
    plt.show()


# === LLAMADO ===
tiempos = bench(10, step=1)


x = [size for size, timing in tiempos]
y = [timing for size, timing in tiempos]


dibujar(x, y)
```
