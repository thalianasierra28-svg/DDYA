## 1) Descripción del problema **[1 pts]**

Una plataforma de red social quiere analizar las conexiones de amistad entre sus usuarios. Cada usuario es un vértice y cada amistad es una arista no dirigida, porque si Ana es amiga de Carlos, entonces Carlos también es amigo de Ana.

El programa debe mostrar con quién está conectado directamente cada usuario, recorrer la red con BFS y DFS desde cualquier usuario, encontrar la conexión con menos saltos entre dos personas y registrar un usuario nuevo conectado con al menos dos usuarios existentes.

Red inicial: 7 usuarios (Ana, Carlos, Laura, Mateo, Sofía, Daniel y Valentina) y 7 amistades:

Ana - Carlos

Ana - Laura

Carlos - Mateo

Carlos - Sofía

Laura - Daniel

Sofía - Valentina

Daniel - Valentina


Se usa una lista de adyacencia porque cada usuario tiene pocas amistades. BFS usa una cola y recorre la red por niveles, por eso sirve para hallar la ruta con menos saltos. DFS usa recursión y avanza por una rama hasta el fondo antes de devolverse.

## 5) Análisis de complejidad **[15 pts]**

 U representa el número de usuarios y E el número de amistades. Cada amistad aparece dos veces en la lista de adyacencia (una por cada amigo), así que todas las listas juntas suman 2A.

Complejidad por función

- add_friendship: O(grado), porque revisa la lista del usuario para no duplicar.
- show_network: O(U + A)
- bfs: O(U + A)
- dfs: O(U + A)
- find_path: O(U + A)
- register_user: O(k · grado), con k amistades nuevas.
- Memoria: O(U + A) para la lista de adyacencia y O(U) extra en BFS y DFS.

Función de BFS

- c1: crear la cola, los visitados y el grado de separación.
- c2 · U: cada usuario sale de la cola una sola vez.
- c3 · 2A: se revisa la lista de amigos de cada usuario, que en total tiene 2A elementos.

T(U, A) = c1 + c2·U + 2·c3·A, que es O(U + A). DFS tiene la misma forma.

Mejor caso (cadena, A = n - 1): T(n) = c1 + c2·n + 2·c3·(n - 1), que es O(n).

Peor caso (red completa, A = n(n - 1)/2): T(n) = c1 + c2·n + c3·n(n - 1), que es O(n²).

En la red (U = 7, A = 7) BFS hace 7 + 2·7 = 21 operaciones, y con Juliana (U = 8, A = 9) hace 8 + 2·9 = 26.


- Mejor caso: f(n) = 3n - 2 (color verde)
- Peor caso: g(n) = n² (color rojo)

## 6) Código funcional **[15 pts]**

```python
from collections import deque
import matplotlib.pyplot as plt

INITIAL_USERS = ["Ana", "Carlos", "Laura", "Mateo", "Sofía", "Daniel", "Valentina"]

INITIAL_FRIENDSHIPS = [
    ("Ana", "Carlos"),
    ("Ana", "Laura"),
    ("Carlos", "Mateo"),
    ("Carlos", "Sofía"),
    ("Laura", "Daniel"),
    ("Sofía", "Valentina"),
    ("Daniel", "Valentina")
]

POSITIONS = {
    "Ana": (0, 3),
    "Carlos": (-1.5, 2),
    "Laura": (1.5, 2),
    "Mateo": (-2.4, 1),
    "Sofía": (-0.6, 1),
    "Daniel": (1.5, 1),
    "Valentina": (0.45, 0),
    "Juliana": (-1.0, 0)
}
OFFSET = (-0.5, -1.3)


def create_initial_network():
    network = {}
    for user in INITIAL_USERS:
        add_user(network, user)
    for a, b in INITIAL_FRIENDSHIPS:
        add_friendship(network, a, b)
    return network


def add_user(network, name):
    if name not in network:
        network[name] = []


def add_friendship(network, a, b):
    if a == b or a not in network or b not in network:
        return False
    if b in network[a]:
        return False
    network[a].append(b)
    network[b].append(a)
    return True


def show_network(network):
    for name in network:
        print(name + ":", ", ".join(network[name]))


def bfs(network, start):
    if start not in network:
        return None, None
    visited = {start}
    queue = deque([start])
    order = []
    degree = {start: 0}
    while queue:
        current = queue.popleft()
        order.append(current)
        for neighbor in network[current]:
            if neighbor not in visited:
                visited.add(neighbor)
                degree[neighbor] = degree[current] + 1
                queue.append(neighbor)
    return order, degree


def dfs(network, start):
    if start not in network:
        return None
    visited = set()
    order = []
    explore(network, start, visited, order)
    return order


def explore(network, current, visited, order):
    visited.add(current)
    order.append(current)
    for neighbor in network[current]:
        if neighbor not in visited:
            explore(network, neighbor, visited, order)


def find_path(network, source, target):
    if source not in network or target not in network:
        return None
    previous = {source: None}
    queue = deque([source])
    while queue:
        current = queue.popleft()
        if current == target:
            break
        for neighbor in network[current]:
            if neighbor not in previous:
                previous[neighbor] = current
                queue.append(neighbor)
    if target not in previous:
        return []
    path = []
    step = target
    while step is not None:
        path.append(step)
        step = previous[step]
    path.reverse()
    return path


def register_user(network, name, friends):
    if name == "" or name in network:
        return "Error: el usuario ya existe o el nombre está vacío."
    friends = list(dict.fromkeys(friends))
    for friend in friends:
        if friend not in network:
            return "Error: " + friend + " no existe en la red."
    if len(friends) < 2:
        return "Error: debe conectarse con al menos dos usuarios existentes."
    add_user(network, name)
    for friend in friends:
        add_friendship(network, name, friend)
    return "Usuario " + name + " registrado con " + str(len(friends)) + " conexiones."


def print_bfs(network, start):
    order, degree = bfs(network, start)
    if order is None:
        print("Error: " + start + " no existe en la red.")
        return
    print("Recorrido BFS desde " + start + ": " + " -> ".join(order))
    for g in range(max(degree.values()) + 1):
        print("  Grado " + str(g) + ": " + ", ".join(x for x in order if degree[x] == g))


def print_dfs(network, start):
    order = dfs(network, start)
    if order is None:
        print("Error: " + start + " no existe en la red.")
        return
    print("Recorrido DFS desde " + start + ": " + " -> ".join(order))


def print_path(network, source, target):
    path = find_path(network, source, target)
    if path is None:
        print("Error: el usuario inicial o el destino no existe en la red.")
    elif path == []:
        print("No existe una conexión entre " + source + " y " + target + ".")
    else:
        print("Ruta encontrada: " + " -> ".join(path))
        print("Grados de separación: " + str(len(path) - 1))
    return path


def draw_network(network, path=None, title="Red"):
    pos = dict(POSITIONS)
    for name in network:
        if name not in pos:
            known = [pos[x] for x in network[name] if x in pos]
            if known:
                x = sum(p[0] for p in known) / len(known) + OFFSET[0]
                y = sum(p[1] for p in known) / len(known) + OFFSET[1]
            else:
                x, y = 2.2, -1
            pos[name] = (x, y)
    path = path or []
    in_path = set(zip(path, path[1:])) | set(zip(path[1:], path))
    fig, ax = plt.subplots(figsize=(8, 6))
    for a in network:
        for b in network[a]:
            if a < b:
                color = "#e65100" if (a, b) in in_path else "#9e9e9e"
                width = 3.5 if (a, b) in in_path else 1.6
                ax.plot([pos[a][0], pos[b][0]], [pos[a][1], pos[b][1]], color=color, lw=width, zorder=1)
    for name in network:
        if path and name == path[0]:
            fill, edge = "#c8e6c9", "#2e7d32"
        elif path and name == path[-1]:
            fill, edge = "#ffcdd2", "#c62828"
        elif name in path:
            fill, edge = "#ffe0b2", "#e65100"
        elif name not in INITIAL_USERS:
            fill, edge = "#e1bee7", "#7b1fa2"
        else:
            fill, edge = "#f8bbd0", "#c2185b"
        ax.text(pos[name][0], pos[name][1], name, ha="center", va="center", fontsize=11, zorder=2,
                bbox=dict(boxstyle="round,pad=0.45", fc=fill, ec=edge, lw=1.8))
    ax.set_title(title, fontsize=13, color="black")
    ax.set_xlim(-3.2, 2.8)
    ax.set_ylim(-1.8, 3.5)
    ax.axis("off")
    plt.show()


network = create_initial_network()
print("Red cargada:", len(network), "usuarios")
```

## 7) Tests aplicados **[15 pts]**
Cada test arma una red nueva con create_initial_network(), para que un test no dependa de otro.

```python
from collections import deque


def bfs_counting(graph, start):
    operations = 0
    visited = {start}
    queue = deque([start])
    while queue:
        current = queue.popleft()
        operations += 1
        for neighbor in graph[current]:
            operations += 1
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)
    return operations


def chain_network(n):
    graph = {i: [] for i in range(n)}
    for i in range(n - 1):
        graph[i].append(i + 1)
        graph[i + 1].append(i)
    return graph


def complete_network(n):
    return {i: [j for j in range(n) if j != i] for i in range(n)}


required = ["create_initial_network", "add_user", "add_friendship", "bfs", "dfs",
            "find_path", "register_user", "print_bfs", "print_dfs", "print_path"]
missing = [f for f in required if f not in globals()]
assert not missing, "Ejecuta primero la celda de Código funcional. Faltan: " + ", ".join(missing)

import io
from contextlib import redirect_stdout

results = []


def check(name, condition):
    results.append(condition)
    print(("OK    " if condition else "FALLA ") + name)


def output_of(function, *args):
    buffer = io.StringIO()
    with redirect_stdout(buffer):
        function(*args)
    return buffer.getvalue()

net = create_initial_network()
check("La red inicial tiene 7 usuarios", len(net) == 7)
check("La red tiene 7 amistades (suma de listas = 2E = 14)", sum(len(x) for x in net.values()) == 14)
check("Ana tiene como amigos a Carlos y Laura", net["Ana"] == ["Carlos", "Laura"])


# BFS
net = create_initial_network()
order, degree = bfs(net, "Ana")
check("BFS desde Ana da el orden esperado", order == ["Ana", "Carlos", "Laura", "Mateo", "Sofía", "Daniel", "Valentina"])
check("BFS calcula bien el grado de separación de cada usuario", degree == {"Ana": 0, "Carlos": 1, "Laura": 1, "Mateo": 2, "Sofía": 2, "Daniel": 2, "Valentina": 3})
check("BFS visita cada usuario una sola vez", len(order) == len(set(order)) == len(net))

# DFS
order = dfs(net, "Ana")
check("DFS desde Ana da el orden esperado", order == ["Ana", "Carlos", "Mateo", "Sofía", "Valentina", "Daniel", "Laura"])
check("DFS visita cada usuario una sola vez", len(order) == len(set(order)) == len(net))

# Ruta mínima con BFS
check("Ruta Ana -> Valentina es la esperada", find_path(net, "Ana", "Valentina") == ["Ana", "Carlos", "Sofía", "Valentina"])
check("La ruta Ana -> Valentina tiene 3 grados de separación", len(find_path(net, "Ana", "Valentina")) - 1 == 3)
check("La ruta de BFS es la mínima (igual al grado del destino)", len(find_path(net, "Ana", "Valentina")) - 1 == bfs(net, "Ana")[1]["Valentina"])


# Registro de un usuario nuevo
net = create_initial_network()
copy_before = {k: list(x) for k, x in net.items()}
check("Rechaza el registro con una sola conexión", register_user(net, "Juliana", ["Mateo"]).startswith("Error") and "Juliana" not in net)
check("Rechaza el registro con un usuario inexistente", register_user(net, "Juliana", ["Mateo", "Pedro"]).startswith("Error") and "Juliana" not in net)
check("Rechaza el mismo usuario repetido como dos conexiones", register_user(net, "Juliana", ["Mateo", "Mateo"]).startswith("Error") and "Juliana" not in net)


# Complejidad
net = create_initial_network()
check("BFS hace V + 2E = 21 operaciones en la red inicial", bfs_counting(net, "Ana") == 21)
register_user(net, "Juliana", ["Mateo", "Daniel"])
check("BFS hace V + 2E = 26 operaciones con Juliana", bfs_counting(net, "Ana") == 26)
check("Mejor caso (cadena): operaciones = 3n - 2", all(bfs_counting(chain_network(n), 0) == 3 * n - 2 for n in [5, 20, 100]))
check("Peor caso (red completa): operaciones = n^2", all(bfs_counting(complete_network(n), 0) == n * n for n in [5, 20, 100]))

print()
print(str(sum(results)) + " de " + str(len(results)) + " tests pasaron")
assert all(results)
```
